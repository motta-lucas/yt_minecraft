```python
from utils.youtube_endpoints import get_playlist_id, get_videos_ids, extract_video_data, save_to_json
from utils.log_config import logs_config
from dotenv import load_dotenv
import requests
import os
import polars as pl
from pathlib import Path
```

```python
load_dotenv("../.env", override=True)

API_KEY = os.getenv("YT_API_KEY")
CHANNEL_HANDLE = ["nico","kiply","skiddzieplays","exnpoo","wudo"]

playlist_id = []

for item in CHANNEL_HANDLE:
    playlist_id.append(get_playlist_id(item,API_KEY))
```

```python
maxResults = 50

videos_ids = []

for item in playlist_id:
    videos_ids = videos_ids + get_videos_ids(maxResults,item,API_KEY)

```

```python
video_data = extract_video_data(videos_ids,API_KEY,maxResults)
```

```python
DATA_DIR = Path.cwd().parent / "data"

save_to_json(video_data,DATA_DIR)
```
